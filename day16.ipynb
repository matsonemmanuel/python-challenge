```python
from datetime import datetime

# Get current datetime
now = datetime.now()

# Extract date and time components
day = now.day
month = now.month
year = now.year
hour = now.hour
minute = now.minute
timestamp = now.timestamp()

# Print results
print(f"Day: {day}")
print(f"Month: {month}")
print(f"Year: {year}")
print(f"Hour: {hour}")
print(f"Minute: {minute}")
print(f"Timestamp: {timestamp}")
```

```python
from datetime import datetime

# Get current datetime
now = datetime.now()

# Format the date and time
formatted_date = now.strftime("%m/%d/%Y, %H:%M:%S")

# Print result
print(formatted_date)
```

```python
from datetime import datetime

# Given date string
date_string = "5 December, 2019"

# Convert string to datetime object
date_object = datetime.strptime(date_string, "%d %B, %Y")

# Print result
print(date_object)
```

```python
from datetime import datetime

# Get the current datetime
now = datetime.now()

# Set the target New Year date (January 1st of the next year)
new_year = datetime(year=now.year + 1, month=1, day=1, hour=0, minute=0, second=0)

# Calculate the time difference
time_difference = new_year - now

# Print the result
print(f"Time left until New Year: {time_difference}")
```

```python
from datetime import datetime

# Define the Unix Epoch time (January 1, 1970)
epoch_time = datetime(1970, 1, 1)

# Get the current datetime
now = datetime.now()

# Calculate the time difference
time_difference = now - epoch_time

# Print the result
print(f"Time difference: {time_difference}")
```
