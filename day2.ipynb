```python
# Data types

first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name
country = "USA"
city = "New York"
age = 25
year = 2021
is_married = False
is_true = True
is_light_on = True
a=b=c=2
print(type(first_name))
print(type(last_name))
print(type(full_name))
print(type(country))
print(type(city))
print(type(age))
print(type(year))
print(type(is_married))
print(type(is_true))
print(type(is_light_on))
print(type(a))
print(type(b))
print(type(c))

#Lenght of the string
print(len(first_name))
print(len(last_name))
print(len(full_name))
#Comparing the length of the string
print(len(first_name) > len(last_name))
print(len(first_name) < len(last_name))
print(len(first_name) == len(last_name))
print(len(first_name) >= len(last_name))
```

```python
num_one = 5
num_two = 4
sum = num_one + num_two  # addition
difference = num_one - num_two # subtraction
product = num_one * num_two  # multiplication
division = num_one / num_two # division
exp = num_one ** num_two # exponent
floor_div = num_one // num_two # floor division
modulus = num_one % num_two # modulus
print(sum)
print(difference)
print(product)
print(division)
print(exp)
print(floor_div)
print(modulus)
```

```python
# The radius of a circle is 30 meters.
# Calculate the area of a circle and the circumference of a circle.
import math
radius = 30
area_of_circle = math.pi * radius ** 2
circumference_of_circle = 2 * math.pi * radius
print(area_of_circle)
print(circumference_of_circle)
```
