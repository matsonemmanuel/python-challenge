```python
import re
from collections import Counter

paragraph = """I love teaching. If you do not love teaching what else can you love. 
I love Python if you do not love something which can give you all the capabilities 
to develop an application what else can you love."""

# Use RegEx to extract words
words = re.findall(r'\b\w+\b', paragraph.lower())

# Count word frequencies
word_counts = Counter(words)

# Find the most frequent word
most_common_word, most_common_count = word_counts.most_common(1)[0]

print(f"The most frequent word is '{most_common_word}' with {most_common_count} occurrences.")
```

```python
import re

# Given text
text = """The position of some particles on the horizontal x-axis are -12, -4, -3 and -1 in the negative direction, 
0 at origin, 4 and 8 in the positive direction."""

# Extract numbers using regex
extracted_numbers = re.findall(r'-?\d+', text)

# Convert to integers
points = list(map(int, extracted_numbers))

# Sort the points
sorted_points = sorted(points)

# Calculate the distance between the two furthest particles
distance = sorted_points[-1] - sorted_points[0]

print(f"Extracted and sorted points: {sorted_points}")
print(f"Distance between the two furthest particles: {distance}")
```

```python
import re
import keyword

def is_valid_variable(var_name):
    # Pattern: Starts with a letter or underscore, followed by letters, digits, or underscores
    pattern = r'^[a-zA-Z_]\w*$'
    
    # Check if it matches the pattern and is not a keyword
    return bool(re.match(pattern, var_name)) and var_name not in keyword.kwlist

# Test cases
print(is_valid_variable('first_name'))  
print(is_valid_variable('first-name'))  
print(is_valid_variable('1first_name'))
print(is_valid_variable('firstname'))   
print(is_valid_variable('_variable'))   
print(is_valid_variable('class')) 
```

```python
import re
from collections import Counter

# Given sentence
sentence = '''%I $am@% a %tea@cher%, &and& I lo%#ve %tea@ching%;. 
There $is nothing; &as& mo@re rewarding as educa@ting &and& @emp%o@wering peo@ple. 
;I found tea@ching m%o@re interesting tha@n any other %jo@bs. 
%Do@es thi%s mo@tivate yo@u to be a tea@cher!?'''

# Function to clean text
def clean_text(text):
    cleaned = re.sub(r'[^a-zA-Z\s]', '', text)  # Remove special characters
    return cleaned.strip()

# Function to find three most frequent words
def most_frequent_words(text):
    words = text.lower().split()  # Convert to lowercase and split
    word_counts = Counter(words)  # Count word occurrences
    return word_counts.most_common(3)  # Get top 3 most common words

# Clean the text
cleaned_text = clean_text(sentence)
print(cleaned_text)

# Get three most frequent words
print(most_frequent_words(cleaned_text))
```
