```python
it_companies = ["Facebook", "Google", "Microsoft", "APPLE", "Intel", "Oracle", "Amazon", "Tesla"]

# Calculate the middle index or indices
length = len(it_companies)

if length % 2 == 1:
    # Odd length, one middle company
    middle_index = length // 2
    removed_company = it_companies.pop(middle_index)
    print(f"Removed middle company: {removed_company}")
else:
    # Even length, two middle companies
    middle_index1 = length // 2 - 1
    middle_index2 = length // 2
    removed_companies = [it_companies.pop(middle_index2), it_companies.pop(middle_index1)]
    print(f"Removed middle companies: {removed_companies}")

print("Updated IT companies list:", it_companies)
```

```python
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]

# Sort the list
ages.sort()

# Find the min and max age
min_age = min(ages)
max_age = max(ages)

print("Sorted ages:", ages)
print("Minimum age:", min_age)
print("Maximum age:", max_age)

# Add the min and max age again to the list
ages.append(min_age)
ages.append(max_age)

print("Updated ages list:", ages)
```

```python
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]

# Sort the list
ages.sort()

# Find the median
length = len(ages)

if length % 2 == 1:
    # Odd length, one middle item
    median = ages[length // 2]
else:
    # Even length, average of the two middle items
    middle1 = ages[length // 2 - 1]
    middle2 = ages[length // 2]
    median = (middle1 + middle2) / 2

print("Median age:", median)
```

```python
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]

# Find the sum of all ages
total_age = sum(ages)

# Find the number of items in the list
num_of_ages = len(ages)

# Calculate the average age
average_age = total_age / num_of_ages

print("Average age:", average_age)
```

```python
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]

# Find the min, max, and average age
min_age = min(ages)
max_age = max(ages)
average_age = sum(ages) / len(ages)

# Calculate the absolute values of (min - average) and (max - average)
min_diff = abs(min_age - average_age)
max_diff = abs(max_age - average_age)

# Print the results
print("Absolute difference between min and average:", min_diff)
print("Absolute difference between max and average:", max_diff)
```

```python
countries = ["USA", "Canada", "Mexico", "Germany", "France", "Italy", "Japan", "China", "India", "Brazil"]

# Find the middle index
length = len(countries)
middle_index = length // 2

if length % 2 == 0:
    # Even length, split equally
    first_half = countries[:middle_index]
    second_half = countries[middle_index:]
else:
    # Odd length, the first half has one more country
    first_half = countries[:middle_index + 1]
    second_half = countries[middle_index + 1:]

# Find the middle country or countries
if length % 2 == 1:
    middle_countries = countries[middle_index]
else:
    middle_countries = countries[middle_index - 1:middle_index + 1]

# Print the results
print("Middle country(ies):", middle_countries)
print("First half of the countries:", first_half)
print("Second half of the countries:", second_half)
```
