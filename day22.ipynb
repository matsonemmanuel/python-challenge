```python
import requests
from bs4 import BeautifulSoup
import json

# URL to scrape
url = 'http://www.bu.edu/president/boston-university-facts-stats/'

# Send a GET request to the URL
response = requests.get(url)
response.raise_for_status()  # Raise an error for bad status codes

# Parse the HTML content
soup = BeautifulSoup(response.content, 'html.parser')

# Dictionary to store the extracted data
bu_facts = {}

# Extracting data based on the structure of the webpage
# Note: The actual HTML structure may vary; adjust the selectors accordingly

# Example: Extracting all sections with their headings and corresponding data
sections = soup.find_all('section')  # Adjust the tag based on actual structure

for section in sections:
    heading = section.find('h2')  # Assuming each section has an h2 heading
    if heading:
        section_title = heading.get_text(strip=True)
        bu_facts[section_title] = {}

        # Extracting key-value pairs within the section
        items = section.find_all(['p', 'li'])  # Adjust tags as necessary
        for item in items:
            # Split the text into key and value based on a delimiter, e.g., ':'
            if ':' in item.text:
                key, value = item.text.split(':', 1)
                bu_facts[section_title][key.strip()] = value.strip()
            else:
                # Handle cases where the text doesn't contain a key-value pair
                bu_facts[section_title][item.text.strip()] = None

# Save the extracted data to a JSON file
with open('bu_facts.json', 'w', encoding='utf-8') as json_file:
    json.dump(bu_facts, json_file, ensure_ascii=False, indent=4)

print("Data has been successfully scraped and saved to 'bu_facts.json'.")
```

```python
import requests
import json

# Send a GET request to the URL
response = requests.get(url)
response.raise_for_status()  # Raise an error for bad status codes

# Parse the HTML content
soup = BeautifulSoup(response.content, 'html.parser')

# Locate the table containing the datasets
table = soup.find('table', {'border': '1'})

if table:
    # Extract table headers
    headers = [header.text.strip() for header in table.find_all('th')]

    # Extract table rows
    rows = []
    for row in table.find_all('tr')[1:]:  # Skip the header row
        cells = row.find_all('td')
        if len(cells) == len(headers):  # Ensure the row has the correct number of cells
            row_data = {headers[i]: cells[i].text.strip() for i in range(len(cells))}
            rows.append(row_data)
else:
    print("Error: Table not found on the webpage.")
    rows = []

# Save the extracted data to a JSON file
with open('uci_datasets.json', 'w', encoding='utf-8') as json_file:
    json.dump(rows, json_file, ensure_ascii=False, indent=4)

print("Data has been successfully scraped and saved to 'uci_datasets.json'.")
```
