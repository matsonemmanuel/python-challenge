```python
numbers = [-4, -3, -2, -1, 0, 2, 4, 6]
filtered_numbers = [num for num in numbers if num <= 0]
print(filtered_numbers)
```

```python
list_of_lists = [[[1, 2, 3]], [[4, 5, 6]], [[7, 8, 9]]]  # making it as one list of lists
flattened_list = [item for sublist1 in list_of_lists for sublist2 in sublist1 for item in sublist2]
print(flattened_list)
```

```python
list_of_tuples = [(i, 1, i, i**2, i**3, i**4, i**5) for i in range(11)]
print(list_of_tuples) # list of tuples with 11 elements
```

```python
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
output = [[country[0][0].upper(), country[0][0][:3].upper(), country[0][1].upper()] for country in countries]
print(output) # list of countries with their capital in upper case and first three letters of the country in upper case
```

```python
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
output = [{'country': country[0][0].upper(), 'city': country[0][1].upper()} for country in countries]
print(output) # list of countries with their capital in upper case and first three letters of the country in upper case
```

```python
names = [[('Emmanuel', 'Matson')], [('David', 'Smith')], [('Donald', 'Trump')], [('Bill', 'Gates')]]
output = [f"{name[0][0]} {name[0][1]}" for name in names]
print(output)
```

```python
slope = lambda x1, y1, x2, y2: (y2 - y1) / (x2 - x1) if x2 != x1 else None
print(slope(1, 2, 3, 4)) # slope of the line between two points (1,2) and (3,4)
```
