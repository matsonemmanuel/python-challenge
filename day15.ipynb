```python
# ModuleNotFoundError

import mathS  # This will raise an error because mathS is not a valid module.
import math  # This will work because math is a valid module.
```

```python
import math
math.PI # This will raise an error because PI is not a valid attribute of the math module.

math.pi # This will work because pi is a valid attribute of the math module.
```

```python
# KeyError

student={'name': 'John', 'age': 25}
student('neme')  # This will raise an error because 'name' is not a valid key in the dictionary.
student['name']  # This will work because 'name' is a valid key in the dictionary.
```

```python
# import error
from math import power  # This will raise an error because power is not a valid function in the math module.
from math import pow  # This will work because pow is a valid function in the math module.
```

```python
# valueError
int('23b') # This will raise an error because '23b' is not a valid integer.

int('23') # This will work because '23' is a valid integer.
```

```python
# ZeroDivisionError

1/0 # This will raise an error because division by zero is not allowed.
1/2 # This will work because division by a non-zero number is allowed.
```
