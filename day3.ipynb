```python
# Calculate the slope, x-intercept and y-intercept of y = 2x -2

# Given line equation: y = 2x - 2

# Coefficients
m = 2        # slope
b = -2       # y-intercept

# Calculate x-intercept (when y = 0)
# 0 = 2x - 2 => 2x = 2 => x = 1
x_intercept = -b / m

# Output the results
print(f"Slope (m): {m}")
print(f"Y-Intercept: (0, {b})")
print(f"X-Intercept: ({x_intercept}, 0)")

```

```python
# Slope is (m = y2-y1/x2-x1). Find the slope and Euclidean distance between point (2, 2) and point (6,10)

import math

# Points
x1, y1 = 2, 2
x2, y2 = 6, 10

# Slope
slope = (y2 - y1) / (x2 - x1)

# Euclidean Distance
distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# Output
print(f"Slope between the points: {slope}")
print(f"Euclidean distance between the points: {distance:.2f}")
```

```python
# Even numbers are divisible by 2 and the remainder is zero. How do you check if a number is even or not using python?

# Function to check if a number is even
def is_even(number):
    if number % 2 == 0:
        return True  # It's even
    else:
        return False  # It's odd

# Test with a number
number = int(input("Enter a number: "))
if is_even(number):
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")
```

```python
# Write a script that prompts the user to enter number of years. Calculate the number of seconds a person can live. Assume a person can live hundred years

# Prompt the user to enter the number of years
years = float(input("Enter number of years: "))

# Assuming a person can live 100 years, calculate the total number of seconds in the given years
# 1 year = 365 days (ignoring leap years), 1 day = 24 hours, 1 hour = 60 minutes, 1 minute = 60 seconds

seconds_in_a_year = 365 * 24 * 60 * 60  # Total seconds in a year
total_seconds_lived = years * seconds_in_a_year  # Total seconds for the given years

# Output the result
print(f"A person living {years} years would have lived {total_seconds_lived:.0f} seconds.")
```
