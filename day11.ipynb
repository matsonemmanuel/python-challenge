```python
def add_all_nums(*args):
    # Check if all arguments are numbers
    if all(isinstance(i, (int, float)) for i in args):
        return sum(args)  # Sum up all numbers
    else:
        return "Error: All inputs must be numbers!"

# Example usage
print(add_all_nums(5, 10, 15))        # Output: 30
print(add_all_nums(3, 7.5, 2, -1))    # Output: 11.5
print(add_all_nums(4, 'hello', 6))    # Output: Error: All inputs must be numbers!
```

```python
def check_season(month):
    month = month.lower()  # Convert to lowercase for case insensitivity

    if month in ["december", "january", "february"]:
        return "Winter"
    elif month in ["march", "april", "may"]:
        return "Spring"
    elif month in ["june", "july", "august"]:
        return "Summer"
    elif month in ["september", "october", "november"]:
        return "Autumn"
    else:
        return "Invalid month"

# Example usage:
month_name = "March"
season = check_season(month_name)
print(f"{month_name} is in {season}.")
```

```python
def calculate_slope(x1, y1, x2, y2):
  if x1==x2:
    return 'Undefined(Vertical line)'
  return (y2-y1)/(x2-x1)
slope=calculate_slope(1,2,3,4)
print(f'The slope of the line is: {slope}')
```

```python
import math

def solve_quadratic_eqn(a, b, c):
    if a == 0:
        return "Not a quadratic equation (a cannot be 0)."

    # Calculate the discriminant
    discriminant = b**2 - 4*a*c

    if discriminant > 0:
        root1 = (-b + math.sqrt(discriminant)) / (2 * a)
        root2 = (-b - math.sqrt(discriminant)) / (2 * a)
        return (root1, root2)  # Two real solutions
    elif discriminant == 0:
        root = -b / (2 * a)
        return (root,)  # One real solution (tuple with one element)
    else:
        real_part = -b / (2 * a)
        imaginary_part = math.sqrt(abs(discriminant)) / (2 * a)
        return (complex(real_part, imaginary_part), complex(real_part, -imaginary_part))  # Two complex solutions

# Example usage:
solutions = solve_quadratic_eqn(1, -3, 2)
print(f"Solutions: {solutions}")
```

```python
def print_list(items):
    for item in items:
        print(item)

my_list = [1, 2, 3, 'apple', 'banana']
print_list(my_list)
```

```python
def reverse_list(arr):
    reversed_arr = []
    for i in range(len(arr) - 1, -1, -1):
        reversed_arr.append(arr[i])
    return reversed_arr

print(reverse_list([1, 2, 3, 4, 5]))
print(reverse_list(["A", "B", "C"]))
```

```python
def capitalize_list_items(items):
    capitalized = []
    for item in items:
        capitalized.append(item.capitalize())
    return capitalized
print(capitalize_list_items(["apple", "banana", "cherry"]))
```

```python
def add_item(lst, item):
    lst.append(item)
    return lst
my_list = [1, 2, 3]
print(add_item(my_list, 4))
```

```python
def remove_item(lst, item):
    if item in lst:
        lst.remove(item)
    return lst
my_list = [1, 2, 3, 4]
print(remove_item(my_list, 3))
```

```python
def sum_of_numbers(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total
print(sum_of_numbers(5))    # Output: 15
print(sum_of_numbers(10))   # Output: 55
print(sum_of_numbers(100))  # Output: 5050
```

```python
def sum_of_odds(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i
    return total
print(sum_of_odds(5))    # Output: 9 (1 + 3 + 5)
print(sum_of_odds(10))   # Output: 25 (1 + 3 + 5 + 7 + 9)
```

```python
def sum_of_even(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 == 0:
            total += i
    return total
print(sum_of_even(5))    # Output: 6 (2 + 4)
print(sum_of_even(10))   # Output: 30 (2 + 4 + 6 + 8 + 10)
```

```python
def evens_and_odds(n):
    evens = 0
    odds = 0
    for i in range(n + 1):
        if i % 2 == 0:
            evens += 1
        else:
            odds += 1
    print(f"The number of evens are {evens}")
    print(f"The number of odds are {odds}")

# Call the function with the desired value
evens_and_odds(100)
  
```
