```python
sentence = "You cannot end a sentence with because because because is a conjunction"
start_index = sentence.find("because")  # Find the start index of the first occurrence
end_index = sentence.rfind("because") + len("because")  # Find the end index of the last occurrence

phrase = sentence[start_index:end_index]  # Slice the string
print(phrase)
```
